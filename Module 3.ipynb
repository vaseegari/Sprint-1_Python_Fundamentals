# Exception Handling

# Definition: 

Exception handling is a way of managing errors in a program so that the program does not stop suddenly. It helps us handle errors properly and show a meaningful message to the user.

# Types of Errors
There are mainly two types of errors:

# 1. Compile-Time Error
A compile-time error occurs before the program starts execution. It usually happens because of a mistake in the code syntax.

Examples:

Missing :

Incorrect indentation

Writing P instead of p

Invalid Python syntax

# 2. Run-Time Error¶

A run-time error occurs while the program is executing. The syntax may be completely correct, but an unexpected situation happens during execution.

# Examples:

Dividing a number by zero

Using an undefined variable

Opening a file that does not exist

Using an invalid list index

# Simple difference:

Compile-time error → Problem found before execution

Run-time error → Problem occurs during execution

# 1.try

# Step 1 – Understand the Concept

try block is used to write code that may cause an exception.

If an error can happen in a particular part of the program, we place that code inside the try block so Python can handle the error instead of stopping the entire program.

# Step 2 – Demonstrate the Concept

# Real-world analogy:

Suppose we are trying to open a locked door. We try to open it first. If something goes wrong, we can handle the situation separately.

Similarly, Python tries to execute the code inside the try block.

# Step 3 – Implement the Concept

```python
try:
    number = 10 / 2
    print("Result:", number)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

# Explanation:

try contains the code that may cause an error.
10 / 2 is successfully executed.
Since no exception occurs, the except block is not executed.

# Concept demonstrated:
The try block allows us to test code that might produce an exception.

# 2. except

# Step 1 – Understand the Concept

except is used to handle an exception that occurs inside the try block.

It prevents the program from terminating suddenly when a known error occurs.

# Step 2 – Demonstrate the Concept

For example, dividing a number by zero is not possible.

Instead of allowing the program to stop, we can catch the error and display a meaningful message.

# Step 3 – Implement the Concept

```python
try:
    number = 10 / 0
except ZeroDivisionError:
    print("Cannot divide a number by zero.")
```

# Explanation:

10 / 0 causes ZeroDivisionError.

except ZeroDivisionError catches that error.

The program displays a friendly message instead of stopping.

# Concept demonstrated:
except handles a specific exception raised inside try.

# 3. else

# Step 1 – Understand the Concept

else block runs only when no exception occurs in the try block.

It is useful when we want to execute some code only after the risky operation has completed successfully.

# Step 2 – Demonstrate the Concept

Think of it like:

Try to complete a task. If there is no problem, do the next step.

# Step 3 – Implement the Concept

```python
try:
    number = 20 / 4
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Division completed successfully.")
    print("Result:", number)
```

# Explanation:

The division is successful.

Therefore, except is skipped.

The else block executes.

# Concept demonstrated:
else executes only when the try block finishes without an exception.

# 4. finally

# Step 1 – Understand the Concept

finally block is used for code that should execute whether an exception occurs or not.

It is commonly used for cleanup operations such as closing files or database connections.

# Step 2 – Demonstrate the Concept

For example, after using a resource, we should close it whether the operation succeeds or fails.

# Step 3 – Implement the Concept

```python
try:
    number = 10 / 2
    print("Result:", number)
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("This block always executes.")
```

# Explanation:

try performs the division.

No exception occurs.

finally still executes.

# Concept demonstrated:
finally runs regardless of whether an exception occurs.

# 5. raise

# Step 1 – Understand the Concept

raise is used when we want to manually generate an exception.

It is useful when the program detects an invalid condition and we want to stop normal execution with a specific error.

#  Step 2 – Demonstrate the Concept

Suppose an age should not be negative.

If the user enters -5, we can manually raise a ValueError.

# Step 3 – Implement the Concept

```python
age = -5

try:
    if age < 0:
        raise ValueError("Age cannot be negative.")
    print("Valid age.")
except ValueError as error:
    print("Error:", error)
```

# Explanation:

We check whether age is negative.

raise ValueError manually creates the exception.

except catches and handles it.

# Concept demonstrated:
raise allows us to create an exception based on our own program logic.

# 6. assert

# Step 1 – Understand the Concept

assert is used to check whether a condition is true.

If the condition is True, the program continues.

If it is False, Python raises an AssertionError.

# Step 2 – Demonstrate the Concept

Suppose a student's mark must be between 0 and 100.

We can use assert to check this condition.

# Step 3 – Implement the Concept

```python
mark = 85

try:
    assert 0 <= mark <= 100, "Mark must be between 0 and 100."
    print("Valid mark.")
except AssertionError as error:
    print("Error:", error)
```

# Explanation:

85 satisfies the condition.

Therefore, assert passes.

The program continues.

# Concept demonstrated:
assert checks assumptions or conditions during program execution.

# 7. Custom Exceptions

# Step 1 – Understand the Concept

A custom exception is an exception that we create ourselves for a specific application requirement.

Python already provides many built-in exceptions, but sometimes our application needs a more meaningful error.

# Step 2 – Demonstrate the Concept

For example, an employee's salary should not be negative.

Instead of using a general ValueError, we can create our own InvalidSalaryError.

# Step 3 – Implement the Concept

```python
class InvalidSalaryError(Exception):
    """Raised when salary is invalid."""
    pass


salary = -1000

try:
    if salary < 0:
        raise InvalidSalaryError("Salary cannot be negative.")
    print("Valid salary.")
except InvalidSalaryError as error:
    print("Error:", error)
```

# Explanation:

InvalidSalaryError is our own exception class.

It inherits from Python's Exception class.

We use raise to generate our custom exception.

except catches and handles it.

# Concept demonstrated:
Custom exceptions allow us to create application-specific error types.

# Built-in Exceptions

Now we will demonstrate the required 10 built-in exceptions.

# 1. ValueError

# Step 1 – Understand the Concept

ValueError occurs when a function receives a value of the correct type but the value itself is not valid for that operation.

# Step 2 – Demonstrate the Concept

Converting "hello" into an integer is not possible.

# Step 3 – Implement the Concept

```python
try:
    number = int("hello")
    print(number)
except ValueError:
    print("Invalid value. Please enter a number.")
```

# What the code demonstrates:

int("hello") produces a ValueError, which is handled using except.

# 2. TypeError
# Step 1 – Understand the Concept

TypeError occurs when an operation is performed on an inappropriate data type.

# Step 2 – Demonstrate the Concept

We cannot add an integer and a string directly.

# Step 3 – Implement the Concept

```python
try:
    result = 10 + "5"
    print(result)
except TypeError:
    print("Cannot add an integer and a string.")
```

# What the code demonstrates:

10 + "5" uses incompatible types, so Python raises TypeError.

# 3. IndexError
# Step 1 – Understand the Concept

IndexError occurs when we try to access an index that does not exist in a sequence such as a list.

# Step 2 – Demonstrate the Concept

If a list contains 3 items, valid indexes are 0, 1, and 2.

Trying to access index 5 causes an error.

# Step 3 – Implement the Concept

```python
numbers = [10, 20, 30]

try:
    print(numbers[5])
except IndexError:
    print("Index does not exist in the list.")
```

# What the code demonstrates:

numbers[5] does not exist, so Python raises IndexError.

# 4. KeyError
# Step 1 – Understand the Concept

KeyError occurs when we try to access a dictionary using a key that does not exist.

# Step 2 – Demonstrate the Concept

Suppose our dictionary contains a person's name and age, but we try to access "city".

# Step 3 – Implement the Concept

```python
person = {
    "name": "Vasee",
    "age": 23
}

try:
    print(person["city"])
except KeyError:
    print("The requested key does not exist.")
```

# What the code demonstrates:

"city" is not available in the dictionary, so Python raises KeyError.

# 5. AttributeError
# Step 1 – Understand the Concept

AttributeError occurs when we try to access a method or attribute that an object does not have.

# Step 2 – Demonstrate the Concept

An integer does not have a lower() method.

# Step 3 – Implement the Concept

```python
number = 100

try:
    print(number.lower())
except AttributeError:
    print("This object does not have the requested attribute or method.")
```

# What the code demonstrates:

number is an integer, so it does not have the lower() method.

# 6. NameError
# Step 1 – Understand the Concept

NameError occurs when we try to use a variable or name that has not been defined.

# Step 2 – Demonstrate the Concept

If we try to print a variable before creating it, Python cannot find that name.

# Step 3 – Implement the Concept

```python
try:
    print(student_name)
except NameError:
    print("The variable has not been defined.")
```

# What the code demonstrates:

student_name was never defined, so Python raises NameError.

# 7. FileNotFoundError
# Step 1 – Understand the Concept

FileNotFoundError occurs when we try to open a file that does not exist at the specified location.

# Step 2 – Demonstrate the Concept

Suppose we try to read data.txt, but the file is not present.

# Step 3 – Implement the Concept

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("The file was not found.")
```

# What the code demonstrates:

Python cannot find data.txt, so FileNotFoundError occurs and is handled.

# 8. ZeroDivisionError
# Step 1 – Understand the Concept

ZeroDivisionError occurs when we try to divide a number by zero.

# Step 2 – Demonstrate the Concept

Mathematically, division by zero is not valid.

# Step 3 – Implement the Concept

```python
try:
    result = 50 / 0
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

# What the code demonstrates:

50 / 0 produces ZeroDivisionError, which is handled by except.

# 9. ImportError
# Step 1 – Understand the Concept

ImportError occurs when Python finds a module but cannot import the requested item from it.

# Step 2 – Demonstrate the Concept

For example, we can try to import a function that does not exist inside a module.

# Step 3 – Implement the Concept

```python
try:
    from math import unknown_function
except ImportError:
    print("The requested item could not be imported.")
```

# What the code demonstrates:

Python finds the math module but cannot find unknown_function inside it, so ImportError occurs

# 10. ModuleNotFoundError
# Step 1 – Understand the Concept

ModuleNotFoundError occurs when Python cannot find the module that we are trying to import.

It is a specific type of ImportError.

# Step 2 – Demonstrate the Concept

If we try to import a module that is not installed or does not exist, Python raises this exception.

# Step 3 – Implement the Concept

```python
try:
    import my_unknown_module
except ModuleNotFoundError:
    print("The requested module was not found.")
```

# What the code demonstrates:

Python cannot find my_unknown_module, so ModuleNotFoundError occurs and is handled.
