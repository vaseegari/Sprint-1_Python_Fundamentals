# 1. What is Logging?

Logging means recording information about what is happening inside a Python program.

It helps us track whether the program is running correctly, where an error occurred, and what happened before the error.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Program started")
logging.warning("Low storage")
logging.error("File not found")
```

# 4.Log Levels

Python logging provides different levels to represent the importance of a message.

The main log levels are:

DEBUG

INFO

WARNING

ERROR

CRITICAL

import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking values")

logging.info("Task completed")

logging.warning("File is empty")

logging.error("File not found")

logging.critical("Program stopped")

# DEBUG

DEBUG is used for detailed information that is useful while developing and debugging a program.

Example:

logging.debug("Checking user input")

It helps developers understand the internal flow of the program.

# INFO

INFO is used to record normal events that happen during program execution.

Example:

logging.info("Task completed")

It tells us that an expected operation has happened successfully.

# WARNING

WARNING indicates that something unexpected happened, but the program can still continue.

Example:

logging.warning("File is empty")

The application does not necessarily stop because of a warning.

# ERROR

ERROR indicates that a particular operation failed.

Example:

logging.error("File not found")

The program may continue running, but that particular operation was unsuccessful.

# CRITICAL

CRITICAL indicates a very serious problem that may prevent the application from continuing.

Example:

logging.critical("Database server is down")

It is used for major failures.

# Easy way to remember
DEBUG     → Detailed debugging information

INFO      → Normal program information

WARNING   → Something unexpected happened

ERROR     → An operation failed

CRITICAL  → Very serious problem

# 5.Formatter

A formatter decides how a log message should look.

For example:

2026-07-10 15:10:00 | INFO | Application started

A formatter makes logs:

readable
    
consistent
    
easier to understand
    
useful for debugging

```python
import logging 
logging.basicConfig(
    level=logging.INFO, 
    format="%(asctime)s | %(levelname)s | %(message)s",
) 
logging.info("Application started")
logging.warning("This is a warning") 
logging.error("This is an error")
```

# Common formatter fields
%(asctime)s     → Date and time

%(levelname)s   → Log level
    
%(message)s     → Log message

%(name)s        → Logger name

# 6.Handlers

A handler decides where the log message should be sent.

The two common handlers are:

Console Handler

File Handler

# Console Handler

A Console Handler sends log messages to the screen or notebook output.

# File Handler

A File Handler saves log messages into a file.

```python
logger = logging.getLogger("logging_example")
logger.setLevel(logging.DEBUG)

formatter = logging.Formatter("%(asctime)s | %(name)s | %(levelname)s | %(message)s")

console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

file_handler = logging.FileHandler("logs/application.log")
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Checking values")
logger.info("Task completed")
logger.warning("File is empty")
logger.error("File not found")
```

# 7.Rotating Logs

Rotating logs automatically create backup log files when the current log file becomes too large.

This is useful for applications that run for a long time and generate many log messages.

# Why is Rotating Logs useful?

It prevents:

huge log files
    
difficult file management
    
excessive storage usage
    
one single log file becoming too large

```python
from logging.handlers import RotatingFileHandler

rot_logger = logging.getLogger("rotation_demo")
rot_logger.setLevel(logging.DEBUG)

rot_handler = RotatingFileHandler("logs/rotating.log",
    maxBytes=200,
    backupCount=3,
)
rot_handler.setFormatter(formatter)
rot_handler.setLevel(logging.DEBUG)

rot_logger.addHandler(rot_handler)

for i in range(1, 25):
    rot_logger.info(f"Log message {i}")
```
