# Control Flow

# 1.if / elif / else

# Defenition

If, elif, and else are conditional statements used to make decisions in a program based on different conditions.

The program checks a condition and executes the code based on whether the condition is True or False.

Simple meaning

if → checks the first condition

elif → checks another condition

else → runs if all conditions are False

```python
age = 20

if age >= 18:          # If this condition is True
    print("Adult")

elif age >= 13:        # If the above condition is False, check this
    print("Teenager")

else:                  # If all conditions are False
    print("Child")
```

# 2.For Loop

# Defenition

A for loop is used to repeat the same code multiple times.

It takes each value one by one from a sequence, such as a list, string, or range()

```python
for i in range(3):       # Repeat 3 times
    print(i)
```

# 3.While Loop

# Defenition

A while loop is used to repeat a block of code as long as a condition is True.

The loop stops when the condition becomes False.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

# 4.Break

# Defenition

break is used to immediately stop a loop, even if the loop condition is still True.

```python
for i in range(1, 6):
    if i == 3:
        break          # Stop the loop when i is 3
    print(i)
```

# Real time example

```python
employees = ["Arun", "Priya", "Vasee", "Kumar"]

for employee in employees:
    if employee == "Vasee":
        break          # Stop when Vasee is found
    print(employee)
```

# 5.Continue

# Defenition

continue is used to skip the current iteration of a loop and move to the next iteration.

It does not stop the loop.

```python
for i in range(1, 6):
    if i == 3:
        continue       # Skip 3
    print(i)
```

# 6.pass

# Defenition

pass is used when you want to leave a block of code empty for now.

It does nothing when executed, but it allows Python code to run without an error.

```python
for i in range(5):
    if i == 3:
        pass          # Do nothing for now
    print(i)
```

# 2.Generators

# Defenition

A generator is a special type of function that produces values one by one, instead of creating all the values at once.

It is useful when working with large amounts of data because it saves memory.

Easy difference:

Normal function → usually returns the result at once

Generator → produces values one by one using yield

```python
def numbers():
    yield 1
    yield 2
    yield 3

for number in numbers():
    print(number)
```

# 3.yield

# Defenition

yield is used inside a generator function to return values one by one.

Unlike return, yield pauses the function and continues from the same place when the next value is requested.

Easy difference:

return → returns a value and ends the function
yield → returns a value and pauses the function, then continues later

```python
def numbers():
    yield 10
    yield 20
    yield 30

for number in numbers():
    print(number)
```

# 4.Decorators

# Defenition

A decorator is used to add extra functionality to an existing function without changing its original code.

Think of it like: function + extra feature = decorated function.

```python
def welcome(func):
    def wrapper():
        print("Starting function")
        func()
        print("Function completed")
    
    return wrapper


@welcome
def greet():
    print("Hello Vasee")


greet()
```

# 5.Closures

# Defenition

A closure is an inner function that remembers the variable from its outer function, even after the outer function has finished.

```python
def outer(name):
    def inner():
        print("Hello", name)

    return inner


greet = outer("Vasee")

greet()
```

# 6.global and nonlocal

# Defenition

Both are keywords used to modify variables from an outer scope, but they work at different levels.

Easy difference

global → modifies a variable from the global scope

nonlocal → modifies a variable from the outer function scope

# global

global is used inside a function to modify a variable that is defined outside the function at the global level.



```python
count = 10

def update():
    global count
    count = 20

update()

print(count)
```

#  nonlocal

nonlocal is used inside a nested function to modify a variable from its outer function.

```python
def outer():
    count = 10

    def inner():
        nonlocal count
        count = 20

    inner()
    print(count)

outer()
```

# Built-in Functions

# 1. map()

# Defenition

map() applies a function to each item in a collection.

```python
numbers = [1, 2, 3, 4]

squares = list(map(lambda x: x * x, numbers))

print(squares)
```

# 2. filter()

# Defenition

filter() is used to select items that satisfy a condition.

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

# 3. zip()

# Defenition

zip() combines elements from two or more collections based on their position.

```python
names = ["Vasee", "Priya", "Arun"]
ages = [23, 24, 25]

employees = list(zip(names, ages))

print(employees)
```

# 4. enumerate()

# Defenition

enumerate() gives both the index and value while looping.

```python
names = ["Vasee", "Priya", "Arun"]

for index, name in enumerate(names):
    print(index, name)
```

# 5. sorted()

# Defenition

sorted() returns the items in sorted order.

```python
numbers = [5, 2, 8, 1, 3]

result = sorted(numbers)

print(result)
```

# 6. any()

# Defenition

any() returns True if at least one condition is True.

```python
numbers = [1, 3, 5, 8]

result = any(num % 2 == 0 for num in numbers)

print(result)
```

# 7. all()

# Defenition

all() returns True only if all conditions are True.

```python
numbers = [2, 4, 6, 8]

result = all(num % 2 == 0 for num in numbers)

print(result)
```
