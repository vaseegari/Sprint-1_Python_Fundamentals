# Object-Oriented Programming

# 1. Class

   # Explanation

A class is a blueprint or template used to create objects. It defines what data an object can have and what actions it can perform.

A class itself is not the actual object. It is only a structure that tells Python how the objects should look and behave.

# Real-time example

In a Student Management System, we can create a Student class to define common student details like name and age.

Example:

Student → Class

Vaseegari → Student information

```python
class Student:
    name = "Vaseegari"
    age = 23

print(Student.name)
print(Student.age)
```

# 2. Object

# Explanation

An object is an actual instance created from a class.

The class is the blueprint, and the object is the actual thing created using that blueprint.

We can create multiple objects from the same class, and each object can have different data.

# Real-time example

In a college system:


Student → Class

student1 → Object

student2 → Object

Both students follow the same structure but can have different information.

```python
class Student:
    name = "Vaseegari"

student1 = Student()

print(student1.name)
```

# 3. Constructor __init__()

# Explanation

A constructor is a special method that runs automatically when an object is created.

In Python, we use __init__() as the constructor.

It is mainly used to initialize the object's data.

# Real-time example

When a new student registers, we need to store their name and age immediately.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Vaseegari", 23)

print(student.name)
print(student.age)
```

# 4. Instance Variables

# Explanation

Instance variables are variables that belong to a particular object.

They are usually created using self.

Each object can have different values for its instance variables.

# Real-time example

Two students can have different names and ages.

```python
class Student:
    def __init__(self, name):
        self.name = name

student1 = Student("Vaseegari")
student2 = Student("Kowsalya")

print(student1.name)
print(student2.name)
```

# 5. Class Variables

# Explanation

A class variable is shared by all objects of a class.

It is defined directly inside the class and outside the methods.

We use it when the same value is common for every object.

# Real-time example

If all students belong to the same college, the college name can be a class variable.

```python
class Student:
    college = "Vels University"

student1 = Student()
student2 = Student()

print(student1.college)
print(student2.college)
```

# 6. Instance Methods

# Explanation

An instance method is a method that works with the data of an object.

It normally uses self.

Instance methods are used when the action depends on a particular object's data.

# Real-time example

A student object can have a display() method to display that student's details

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)

student = Student("Vaseegari")
student.display()
```

# 7. Static Methods

# Explanation

A static method is a method that does not need object data or class data.

It does not use self or cls.

We use the @staticmethod decorator.

Static methods are useful for utility operations related to a class.

# Real-time example

A calculator's addition operation does not need any student or object information.

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

# 8. Class Methods

# Explanation

A class method works with the class itself instead of a particular object.

It uses cls as the first parameter.

We use the @classmethod decorator.

Class methods are useful when we need to access or modify class-level data.

# Real-time example

The college name is common to all students, so we can access it using a class method.

```python
class Student:
    college = "Vels University"

    @classmethod
    def show_college(cls):
        print(cls.college)

Student.show_college()
```

# 9. Properties @property

# Explanation

A property allows us to access a method like an attribute or variable.

Normally, a method is called using ().

With @property, we can access it without ().

Properties are useful for controlling access to data.

# Real-time example

A user can access their account balance without directly accessing the internal variable.

```python
class Student:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

student = Student("Vaseegari")

print(student.name)
```

# 10. Encapsulation

# Explanation

Encapsulation means keeping data and the methods that work with that data together inside a class.

It also helps control access to important data.

Python uses naming conventions such as _ and __ for protected/private-style data.

# Real-time example

In a banking system, the account balance should not be freely changed by anyone. We can control access through methods.

```python
class BankAccount:
    def __init__(self):
        self.__balance = 5000

    def show_balance(self):
        print(self.__balance)

account = BankAccount()

account.show_balance()
```

# 11. Abstraction

# Explanation

Abstraction means hiding unnecessary internal details and showing only the required functionality.

The user only needs to know what the system does, not how it works internally.

# Real-time example

When using an ATM, we select "Withdraw Money". We do not need to know how the bank server processes the transaction internally.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):

    def sound(self):
        print("Bark")

dog = Dog()
dog.sound()
```

# 12. Inheritance

# Explanation

Inheritance allows a child class to reuse the properties and methods of a parent class.

The main purpose of inheritance is code reusability.

# Real-time example

A Dog is an Animal, so the Dog class can use common Animal functionality.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    pass

dog = Dog()
dog.eat()
```

# 13. Multiple Inheritance

# Explanation

Multiple inheritance means one child class inherits from more than one parent class.

This allows the child to use features from multiple parent classes.

# Real-time example

A child may get different skills from both parents.

Father → Driving

Mother → Cooking

Child → Both

```python
class Father:
    def drive(self):
        print("Driving")

class Mother:
    def cook(self):
        print("Cooking")

class Child(Father, Mother):
    pass

child = Child()

child.drive()
child.cook()
```

# 14. Multilevel Inheritance

# Explanation

Multilevel inheritance means inheritance happens across multiple levels.

One class inherits from another class, and another class inherits from that child.

# Real-time example

Animal

   ↓
   
Dog

   ↓
   
Puppy


Puppy can use methods from both Dog and Animal.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

class Puppy(Dog):
    pass

puppy = Puppy()

puppy.eat()
puppy.bark()
```

# 15. Hierarchical Inheritance

# Explanation

Hierarchical inheritance means multiple child classes inherit from the same parent class.

Real-time example

       Animal
       
       /    \
     Dog    Cat

Both Dog and Cat can use common Animal functionality.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    pass

class Cat(Animal):
    pass

dog = Dog()
cat = Cat()

dog.eat()
cat.eat()
```

# 16. Polymorphism

# Explanation

Polymorphism means "many forms."

The same method name can perform different actions depending on the object.

# Real-time example

Both Dog and Cat have a sound() method.

Dog → Bark

Cat → Meow

The method name is the same, but the behavior is different.

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

# 17. Method Overriding

# Explanation

Method overriding happens when a child class provides its own implementation of a method that already exists in the parent class.

It allows the child class to change the parent's behavior.

# Real-time example

The parent class says every animal makes a sound, but Dog provides its own specific sound.

```python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

dog = Dog()
dog.sound()
```

# 18. Method Overloading – Python Approach

# Explanation

Method overloading means using the same method name with different numbers or types of arguments.

Python does not support traditional method overloading like Java.

Instead, we commonly use default arguments or *args.

# Real-time example

A calculator can add one number or two numbers using the same method.

```python
class Calculator:
    def add(self, a, b=0):
        return a + b

calc = Calculator()

print(calc.add(10))
print(calc.add(10, 20))
```

# 19. Composition

# Explanation

Composition means one class contains an object of another class.

It represents a strong has-a relationship.

The contained object is created as part of the main object's structure.

# Real-time example

A Car has an Engine.

Car → Engine

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

car = Car()

car.engine.start()
```

# 20. Aggregation

# Explanation

Aggregation is also a "has-a" relationship, but the objects can exist independently.

The main class uses an existing object instead of creating it internally.

# Real-time example

A College has Students, but a Student can exist even if the College object is removed.

```python
class Student:
    def __init__(self, name):
        self.name = name

class College:
    def __init__(self, student):
        self.student = student

student = Student("Vaseegari")
college = College(student)

print(college.student.name)
```

# 21. Association

# Explanation

Association means two classes are connected and interact with each other.

Neither class necessarily owns the other.

It represents a general relationship between objects.

# Real-time example

A Teacher teaches a Student.

Teacher and Student can exist independently.

```python
class Teacher:
    def teach(self):
        print("Teaching")

class Student:
    def learn(self, teacher):
        teacher.teach()
        print("Learning")

teacher = Teacher()
student = Student()

student.learn(teacher)
```

# 22. Magic / Dunder Methods

# Explanation

Dunder means double underscore.

Dunder methods are special methods provided by Python. They start and end with double underscores.

Examples:

__init__()

__str__()

__repr__()

__len__()

Python automatically calls these methods in specific situations.

# Real-time example

When we use:

print(object)

Python can use __str__().

When we use:

len(object)

Python can use __len__().

# 23. super()

# Explanation

super() is used to access a method or constructor from the parent class.

It is especially useful when the child class overrides a parent method but still needs the parent's functionality.

# Real-time example

A child class can first perform the parent's operation and then perform its own operation.

```python
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        super().show()
        print("Child method")

child = Child()

child.show()
```

# 24. __str__()

# Explanation

__str__() defines the user-friendly string representation of an object.

When we use print(object), Python calls __str__().

It is useful when we want to display meaningful information instead of the default object representation.

# Real-time example

Instead of showing a technical object address, we can show:

Student: Vaseegari

```python
class Student:
    def __str__(self):
        return "Student: Vaseegari"

student = Student()

print(student)
```

# 25. __repr__()


# Explanation

__repr__() provides a more developer-friendly representation of an object.

It is mainly useful for debugging and understanding what an object contains.

# Real-time example

For a Student object, we may want to see:

Student('Vaseegari')

instead of a generic object address.

```python
class Student:
    def __repr__(self):
        return "Student('Vaseegari')"

student = Student()

print(repr(student))
```

# 26. __len__()

# Explanation

__len__() defines what Python should return when we use len() on an object.

Normally, len() works with strings, lists, tuples, etc. By defining __len__(), we can make it work with our own objects.

# Real-time example

A Team object contains three members. We can use len(team) to find the number of members.

```python
class Team:
    def __init__(self):
        self.members = ["A", "B", "C"]

    def __len__(self):
        return len(self.members)

team = Team()

print(len(team))
```

# Bonus Topics

# 27. Dataclasses

# Explanation

A dataclass is a special type of class mainly used for storing data.

Normally, when creating a class, we may need to write __init__(), __repr__(), and other methods manually.

A dataclass can automatically provide several of these common methods.

# Real-time example

An Employee object may only need to store:

Name

Age

Department

A dataclass is useful for this type of data.

```python
from dataclasses import dataclass

@dataclass
class Employee:
    name: str
    age: int

employee = Employee("Vaseegari", 23)

print(employee)
```

# 28. Singleton Pattern

# Explanation

The Singleton Pattern ensures that only one instance of a class is created.

If the program tries to create the object again, it returns the same object instead of creating a new one.

# Real-time example

A large application may need only one:

Configuration manager

Logger

Shared database connection

Instead of creating multiple objects, we can maintain one shared instance.

```python
class Singleton:
    instance = None

    def __new__(cls):
        if cls.instance is None:
            cls.instance = super().__new__(cls)
        return cls.instance

a = Singleton()
b = Singleton()

print(a is b)
```
