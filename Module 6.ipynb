# 1.Database Design
# What is a Database?

A database is an organized collection of data that allows us to store, manage, and retrieve information easily.

**For example, a college application can store:**

Student details

Department details

Course details

Marks

We can use SQL to work with the data stored in a database.

# What is a Table?

A table stores data in rows and columns.

**Example:**

students



        id      name          age        department 

        1       Vaseegari     23            IT 
 
        2       Priya         22            CSE   
 
        3       Arun          24            IT         



Column → represents a type of information

Row → represents one record


# Creating a Database and Tables

**Example using MySQL:**

CREATE DATABASE college;


USE college;


Create a department table:


CREATE TABLE departments (

    department_id INT PRIMARY KEY,
    
    department_name VARCHAR(50)
    
);

Create a student table:


CREATE TABLE students (

    student_id INT PRIMARY KEY,
    
    student_name VARCHAR(50),
    
    age INT,
    
    department_id INT,
    
    FOREIGN KEY (department_id)
    
        REFERENCES departments(department_id)
        
);


# Primary Key
# What is a Primary Key?

A primary key is a column that uniquely identifies each row in a table.

  **A primary key:**

      must be unique

      cannot contain NULL

      identifies a specific record

**Example:**

CREATE TABLE students (

    student_id INT PRIMARY KEY,
    
    student_name VARCHAR(50),
    
    age INT
    
);



**Here:**

student_id → Primary Key

**Example data:**

1 → Vaseegari

2 → Priya

3 → Arun

Each student has a unique ID.

# Foreign Key
# What is a Foreign Key?

A foreign key is a column that creates a connection between two tables.

It refers to the primary key of another table.

**Example:**

CREATE TABLE departments (

    department_id INT PRIMARY KEY,
    
    department_name VARCHAR(50)
    
);

CREATE TABLE students (

    student_id INT PRIMARY KEY,
    
    student_name VARCHAR(50),
    
    department_id INT,

    FOREIGN KEY (department_id)
    
        REFERENCES departments(department_id)
        
);


Here:

departments.department_id

            ↑
            |
students.department_id


The foreign key connects students with departments.

# Relationships

Relationships describe how tables are connected.

**The common relationships are:**

One-to-One

One-to-Many

Many-to-Many

# One-to-One

One record in Table A is connected to one record in Table B.

**Example:**

Person → Passport


One person has one passport.

# One-to-Many

One record in one table can be connected to many records in another table.

**Example:**

Department

    ↓
Students

One department can have many students.

# Many-to-Many

Many records in one table can be connected to many records in another table.

**Example:**

Students ↔ Courses

One student can take many courses, and one course can have many students.

Usually, a third table called a junction table is used.

# Constraints
# What are Constraints?

Constraints are rules applied to table columns to control the data stored in the database.

**Common constraints:**

PRIMARY KEY

FOREIGN KEY

NOT NULL

UNIQUE

DEFAULT

CHECK

# NOT NULL

The column cannot have an empty value.

CREATE TABLE students (

    student_id INT PRIMARY KEY,
    
    student_name VARCHAR(50) NOT NULL
    
);

# UNIQUE

 All values in the column must be different.

   email VARCHAR(100) UNIQUE

# DEFAULT

Assigns a default value if no value is provided.

  status VARCHAR(20)   DEFAULT 'Active'

# CHECK

Checks whether a condition is satisfied.

age INT CHECK (age >= 18)

# 2.Queries

# SELECT
# What is SELECT?

SELECT is used to retrieve data from a table.


**Example:**

  SELECT * FROM students;
  
**Output**

    1 | Vaseegari | 23 | 1

    2 | Priya     | 22 | 2

    3 | Arun      | 24 | 1

    4 | Karthik   | 21 | 3

    5 | Divya     | 23 | 2

To select specific columns:

SELECT student_name, age
FROM students;

# WHERE
# What is WHERE?

WHERE is used to filter records based on a condition.

**Example:**

SELECT *
FROM students
WHERE age > 22;

**Output:**


   1 | Vaseegari | 23 | 1

   3 | Arun      | 24 | 1

   5 | Divya     | 23 | 2

Only students whose age is greater than 22 are returned.

# ORDER BY
# What is ORDER BY?

ORDER BY is used to sort the result.



**Ascending**


SELECT *
FROM students
ORDER BY age ASC;


**Descending**


SELECT *
FROM students
ORDER BY age DESC;



**Example descending output:**

      Arun       | 24

      Vaseegari  | 23

      Divya      | 23
 
      Priya      | 22

      Karthik    | 21

# GROUP BY
# What is GROUP BY?

GROUP BY groups rows that have the same value.

It is commonly used with aggregate functions such as:

COUNT()

SUM()

AVG()

MIN()

MAX()


**Example:**

SELECT department_id, COUNT(*) AS student_count
FROM students
GROUP BY department_id;


**Output**

    department_id | student_count
    1             | 2
    2             | 2
    3             | 1

This tells us how many students are in each department.

# HAVING
# What is HAVING?

HAVING is used to filter grouped results.

WHERE filters rows before grouping.

HAVING filters groups after grouping.


**Example:**

SELECT department_id, COUNT(*) AS student_count
FROM students
GROUP BY department_id
HAVING COUNT(*) > 1;


**Output:**

    department_id | student_count
    1             | 2
    2             | 2

Only departments with more than one student are displayed.

# DISTINCT
# What is DISTINCT?

DISTINCT removes duplicate values from the result.

**Example:**

SELECT DISTINCT age
FROM students;

**Output:**

21

22

23

24

If multiple students have the same age, the age appears only once.

# LIMIT
# What is LIMIT?

LIMIT restricts the number of records returned.

**Example:**

SELECT *
FROM students
LIMIT 3;

**Output:**
 
    1 | Vaseegari | 23 | 1
    2 | Priya     | 22 | 2
    3 | Arun      | 24 | 1

Only the first 3 records are returned.

# 3.Joins
# What is a JOIN?

A JOIN is used to combine data from two or more related tables.

**We have:**

students

departments

They are connected using:

students.department_id

        ↓
departments.department_id

# INNER JOIN
# What is INNER JOIN?

INNER JOIN returns only the records that have matching values in both tables.

**Example:**

SELECT
    students.student_name,
    departments.department_name
    
FROM students
INNER JOIN departments

ON students.department_id = departments.department_id;


**Output**

    Vaseegari | IT
    Priya     | CSE
    Arun      | IT
    Karthik   | ECE
    Divya     | CSE

Only matching department records are returned.

# LEFT JOIN
# What is LEFT JOIN?

**LEFT JOIN returns:**

all records from the left table

matching records from the right table

If there is no match, the right-side value becomes NULL.

**Example:**

SELECT
    students.student_name,
    departments.department_name
    
FROM students
LEFT JOIN departments

ON students.department_id = departments.department_id;


All students are returned, even if a matching department is not available.

# RIGHT JOIN
# What is RIGHT JOIN?

**RIGHT JOIN returns:**

all records from the right table

matching records from the left table

**Example:**

SELECT
    students.student_name,
    departments.department_name
    
FROM students
RIGHT JOIN departments

ON students.department_id = departments.department_id;


This is useful when we want every department to appear, even if it has no students.

**Note:** MySQL supports RIGHT JOIN, but support for RIGHT JOIN varies by database system

# What is FULL JOIN?

**FULL JOIN returns:**

all matching records

all unmatched records from the left table

all unmatched records from the right table

**Conceptually:**

Left table + Matching records + Right table

**Standard SQL syntax:**

SELECT
    students.student_name,
    departments.department_name
    
FROM students
FULL OUTER JOIN departments

ON students.department_id = departments.department_id;



Important: MySQL does not directly support FULL OUTER JOIN. In MySQL, it can commonly be achieved using LEFT JOIN, RIGHT JOIN, and UNION.

# CROSS JOIN
# What is CROSS JOIN?

CROSS JOIN returns every possible combination of rows from two tables.

If:

Table A → 3 rows

Table B → 2 rows

Then:

3 × 2 = 6 rows

**Example:**

SELECT
    students.student_name,
    departments.department_name
    
FROM students

CROSS JOIN departments;

Every student is combined with every department.

# 4.Advanced SQL

# UNION
# What is UNION?

UNION combines the results of two SELECT queries and removes duplicate rows.

**Code**

SELECT student_name

FROM students

WHERE department_id = 1


UNION

SELECT student_name

FROM students

WHERE department_id = 2;

**Output**


    Vaseegari

    Arun
 
    Priya

    Divya

The two result sets are combined into one.

**Important**

Both SELECT statements should return the same number of columns with compatible data types.

# CASE
# What is CASE?

CASE is used to perform conditional logic in SQL.

It is similar to if-elif-else in Python.


**Code**

SELECT
    student_name,
    
    age,
    
    CASE
    
        WHEN age >= 23 THEN 'Senior'
        
        ELSE 'Junior'
        
    END AS student_level
    
FROM students;


**Output**

     Vaseegari | 23 | Senior
     Priya     | 22 | Junior
     Arun      | 24 | Senior
     Karthik   | 21 | Junior
     Divya     | 23 | Senior

# Subqueries
# What is a Subquery?

A subquery is a query written inside another SQL query.


**Example:**

Find students whose age is greater than the average age.

SELECT student_name, age

FROM students

WHERE age > (

    SELECT AVG(age)
    
    FROM students
    
);


The inner query:

SELECT AVG(age)

FROM students;


calculates the average age.


The outer query uses that result to find students above the average.

# Common Table Expressions (CTEs)
# What is a CTE?

A CTE is a temporary named result that makes complex queries easier to read and manage.

It uses the WITH keyword.

**Code**

WITH student_data AS (

    SELECT *
    
    FROM students
    
    WHERE age >= 23
    
)

SELECT *
FROM student_data;


**Output**

     1 | Vaseegari | 23 | 1
     3 | Arun      | 24 | 1
     5 | Divya     | 23 | 2

     
# Why use CTEs?

**CTEs make complex queries:**

easier to read

easier to understand

easier to maintain

easier to reuse within a query

# Views
# What is a View?

A view is a virtual table based on the result of a SQL query.

It does not normally store a separate copy of the underlying data.

# Create a View
CREATE VIEW student_department_view AS

SELECT

    students.student_name,
    
    departments.department_name
    
FROM students

INNER JOIN departments

ON students.department_id = departments.department_id;


**Now we can query the view:**

SELECT *
FROM student_department_view;


**Output**

    Vaseegari | IT
    Priya     | CSE
    Arun      | IT
    Karthik   | ECE
    Divya     | CSE


# Why use Views?

**Views can:**

simplify complex queries 

provide reusable query results

restrict access to selected columns

make reporting easier

# Indexes
# What is an Index?

An index is a database structure that helps the database find rows faster.

It is similar to an index in a book.

Instead of checking every page, we can use the index to find information quickly.

**Code**

CREATE INDEX idx_student_name

ON students(student_name);

Now the database can use this index when appropriate for searches involving 

student_name.



**Important**

Indexes can improve read/search performance, but they also require storage and can add overhead when inserting or updating data.

# Window Functions
# What are Window Functions?

Window functions perform calculations across a set of related rows without combining those rows into one result row.

**They are useful for:**

ranking

numbering rows

comparing previous and next rows

calculating running values



**Common window functions include:**

ROW_NUMBER()

RANK()

DENSE_RANK()

LAG()

LEAD()

# ROW_NUMBER()
# What is ROW_NUMBER()?

ROW_NUMBER() gives a unique sequential number to each row.

**Code**

SELECT

    student_name,
    
    age,
    
    ROW_NUMBER() OVER (ORDER BY age DESC) AS row_num
    
FROM students;

**Output**

    Arun      | 24 | 1
    Vaseegari | 23 | 2
    Divya     | 23 | 3
    Priya     | 22 | 4
    Karthik   | 21 | 5

Even when two students have the same age, they get different row numbers.

# RANK()
# What is RANK()?

RANK() assigns the same rank to tied values.

It leaves a gap after a tie.

**Code**

SELECT

    student_name,
    
    age,
    
    RANK() OVER (ORDER BY age DESC) AS student_rank
    
FROM students;

**Output**

    Arun      | 24 | 1
    Vaseegari | 23 | 2
    Divya     | 23 | 2
    Priya     | 22 | 4
    Karthik   | 21 | 5

Notice that after rank 2, the next rank is 4.

# DENSE_RANK()
# What is DENSE_RANK()?

DENSE_RANK() also gives the same rank to tied values, but it does not leave a gap.

**Code**

SELECT

    student_name,
    
    age,
    
    DENSE_RANK() OVER (ORDER BY age DESC) AS student_rank
    
FROM students;


**Output**

     Arun      | 24 | 1
     Vaseegari | 23 | 2
     Divya     | 23 | 2
     Priya     | 22 | 3
     Karthik   | 21 | 4

     
**RANK vs DENSE_RANK**

     Values:       24   23   23   22

     RANK:         1    2    2    4
     DENSE_RANK:    1    2    2    3

# LAG()
# What is LAG()?

LAG() allows us to access a value from a previous row.

It is useful when comparing the current row with the previous row.

**Code**

SELECT
    student_name,
    
    age,
    
    LAG(age) OVER (ORDER BY age) AS previous_age
    
FROM students;

**Example Output**

     Karthik   | 21 | NULL
     Priya     | 22 | 21
     Vaseegari | 23 | 22
     Divya     | 23 | 23
     Arun      | 24 | 23

For each row, LAG() gives the age from the previous row.

# LEAD()
# What is LEAD()?

LEAD() allows us to access a value from the next row.

**Code**

SELECT

    student_name,
    
    age,
    
    LEAD(age) OVER (ORDER BY age) AS next_age
    
FROM students;

**Example Output**

     Karthik   | 21 | 22
     Priya     | 22 | 23
     Vaseegari | 23 | 23
     Divya     | 23 | 24
     Arun      | 24 | NULL

For each row, LEAD() gives the age from the next row.
