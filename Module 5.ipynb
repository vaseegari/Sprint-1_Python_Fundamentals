# Module 5 - FILE HANDLING

# What is File Handling?

File handling is the process of creating, reading, writing, updating, and managing files and folders using Python.

It allows a program to store data permanently instead of keeping it only in memory.

Python can work with different file types such as:

Text files

CSV files

JSON files

Pickle files

# Why is File Handling used?

File handling helps us:

store data permanently

read existing data

create new files

update files

save application results

manage files and folders

exchange data between applications

For example, an application can save user information in a JSON file and read it later.

# 1. Reading Text Files
# What is Reading a File?

Reading a file means getting the data that is already stored inside a file.

Python uses the open() function to open a file. We use "r" mode when we want to read a file.

# Why do we use it?

We use file reading when we need to retrieve stored information, such as user data, reports, logs, or configuration details.

# Methods:

read()      - Reads entire file

readline()  - Reads one line

readlines() - Reads all lines as a list

# 2. Writing Text Files
# What is Writing a File?

Writing means storing new information inside a file.

We use "w" mode for writing.

# Why do we use it?

It is useful when we want to create a new file or replace the existing content of a file.

Modes:

w - Write (overwrite)

a - Append(add at the end)

x - Create new file

```python
sampletxt = open("sample.txt", "w")
sampletxt.write("Hello World.\n")
sampletxt.write("This is Aishu.\n")
sampletxt.write("This file is my Sprint 1 project.")
sampletxt.close()

print("File created")
```

```python
sampletxt = open("sample.txt", "r")
data = sampletxt.read()
print(data)
sampletxt.close()
```

# 3. CSV Files
# What is CSV?

CSV means Comma-Separated Values.

It is commonly used to store table-like data.

Example:

     Name,Age,City
  
     Vaseegari,23,Chennai

     Priya,22,Madurai

# Why do we use CSV?

CSV is useful for:

student records

employee data

sales data

reports

datasets

```python
#Writting CSV

import csv

data = [
    ["Name", "Age", "City"],
    ["Vaseegari", 23, "Chennai"],
    ["Priya", 22, "Madurai"]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created successfully.")
```

```python
#Reading CSV

import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

# 4. JSON Files
# What is JSON?

JSON stands for JavaScript Object Notation.

It is a commonly used format for storing and exchanging structured data.

Example:

       {
             "name": "Vaseegari",
    
             "age": 23,
    
             "city": "Chennai"
       }


# Why do we use JSON?

JSON is commonly used in:

APIs

web applications

configuration files

data exchange

```python
# Writting JSON

import json

student = {
    "name": "Vaseegari",
    "age": 23,
    "city": "Chennai"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)

print("JSON file created successfully.")
```

```python
# Reading JSON

import json

with open("student.json", "r") as file:
    data = json.load(file)

print(data)
```

# 5. Pickle Files
# What is Pickle?

Pickle is a Python module used to store Python objects in a binary format.

It can store:

list
    
dictionary
    
tuple
    
Python objects
    
# Why do we use Pickle?

Pickle is useful when we want to save Python objects and load them again later.

```python
# Writing Pickle

import pickle

data = ["Python", "SQL", "Django"]

with open("skills.pkl", "wb") as file:
    pickle.dump(data, file)

print("Pickle file created.")
```

```python
# Reading Pickle

import pickle

with open("skills.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

# 6. Context Manager – with
# What is a Context Manager?

A context manager manages resources automatically.

When working with files, the with statement automatically closes the file after the operation is completed.

# Without with

file = open("sample.txt", "r")

data = file.read()

file.close()

We have to manually close the file.

# With with

with open("sample.txt", "r") as file:

    data = file.read()
    

print(data)


The file is automatically closed.


# Why do we use it?

It:

automatically closes files

prevents resource leaks

makes code cleaner

safely manages resources


```python
# to write a file

with open("sample1.txt", "w") as file:
    file.write("Hello World")
    file.write("\nThis is Aishu")

print("File created successfully.")
```

```python
# to read a file

with open("sample1.txt", "r") as file:
    data = file.read()

print(data)
```

# 9. shutil
# What is shutil?

shutil is used for high-level file and directory operations.

It can copy, move, and remove files and directories.

```python
#Copy a File

import shutil

shutil.copy("sample.txt", "backup.txt")

print("File copied successfully.")
```

```python
#Move a File

import shutil

shutil.move("backup.txt", "outputs/backup.txt")

print("File moved successfully.")
```

# 10. File Validation
# What is File Validation?

File validation means checking whether a file exists and whether it is the correct type before processing it.

# Why do we use it?

It prevents errors when a program tries to process a missing or invalid file.

```python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist.")
elif not file_path.is_file():
    print("Path is not a file.")
else:
    print("File is valid.")
```

```python
#Check File Extension

from pathlib import Path

file_path = Path("students.csv")

if file_path.suffix == ".csv":
    print("Valid CSV file.")
else:
    print("Invalid file type.")
```

# 11. Directory Operations
# What are Directory Operations?

Directory operations are actions performed on folders, such as:

creating a folder

checking a folder

listing files

copying a folder

moving a folder

```python
#Create Directory

from pathlib import Path

folder = Path("data")

folder.mkdir(exist_ok=True)

print("Directory created.")
```

```python
#List Files

from pathlib import Path

folder = Path("data")

for file in folder.iterdir():
    print(file)
```

```python
#Check Directory

from pathlib import Path

folder = Path("data")

if folder.is_dir():
    print("Directory exists.")
else:
    print("Directory does not exist.")
```
